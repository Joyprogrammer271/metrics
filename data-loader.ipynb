```python
import torchvision
import torchvision.transforms as transforms
import numpy as np

def load_cifar10(data_path="./data", train=False):
    """
    加载 CIFAR-10 数据集，返回 numpy 格式
    train: True 加载训练集（50000张），False 加载测试集（10000张）
    """
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
    ])

    dataset = torchvision.datasets.CIFAR10(
        root=data_path,
        train=train,
        download=True,
        transform=transform
    )

    # CIFAR-10 的数据在 dataset.data 和 dataset.targets 里
    images = dataset.data  # [N, 32, 32, 3]
    labels = np.array(dataset.targets)  # [N]

    return images, labels


# 测试代码
if __name__ == "__main__":
    print("正在下载/加载 CIFAR-10 测试集...")
    images, labels = load_cifar10(train=False)
    print(f"加载成功！图片数量: {len(images)}，标签数量: {len(labels)}")
    print(f"图片形状: {images.shape}")  # 应该是 (10000, 32, 32, 3)
```
